```python
from geomloss import SamplesLoss
def ot_dist_geomloss(x, y, blur=.05):
    x = x.cuda()
    y = y.cuda()
    loss = SamplesLoss(loss="sinkhorn", p=2, blur=blur, debias=False)
    return torch.sqrt(2 * loss(x, y)).item()
```

```python
import ot
def ot_dist(X, Y, ot_epsilon = 0.1):
    a = torch.ones(X.shape[0]) / X.shape[0]
    b = torch.ones(Y.shape[0]) / Y.shape[0]
    M = ot.dist(X, Y)
    dist = ot.sinkhorn2(a, b, M, ot_epsilon, method='sinkhorn_log')
    return torch.sqrt(dist).item()
```

```python
x = torch.randn(1000, 50)
y = torch.randn(2000, 50)
```

```python
print(ot_dist_geomloss(x,y, 0.05), ot_dist(x, y, 0.05**2))
```
